```python
df["year"] = df["posting_date"].dt.year
```

```python
df["month"] = df["posting_date"].dt.month
```

```python
df["job_title"].head()
```

```python
job_month = (
    df.groupby(["job_title", "month"])
      .size()
      .reset_index(name="count")
)
```

```python
stacked_df = job_month.pivot(
    index="job_title",
    columns="month",
    values="count"
).fillna(0)
```

```python
stacked_df.plot(
    kind="bar",
    stacked=True,
    figsize=(12,6)
)

plt.title("Month-wise Job Postings by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)
plt.legend(title="Month", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```
